```python
class Car:
    """A simple attempt to model a Car."""
    def __init__(self, name, colour):
        """Initializing name, colour and model attributes."""
        self.name = name
        self.colour = colour
        
    def get_Speed(self):
        """Setting the speed of the car."""
        print("Top speed of " + self.name + " is 150 km/hr") 
```

```python
# Creating an instance/object called Honda_City
Honda_City = Car("Honda City","Red")

# Calling the method using the name of the instance
Honda_City.get_Speed() 
```

```python
class Car:
	def setDetails(self, model, regno):
		self.model = model
		self.regno = regno
	def getModel(self):
		return self.model
	def getRegno(self):
		return self.regno
    
Hyundai = Car()

Maruthi = Car()

model1 = input('car1 model: ')
regno1 = input('car1 regno: ')

model2 = input('car2 model: ')
regno2 = input('car2 regno: ')

Hyundai.setDetails(model1, regno1)
Maruthi.setDetails(model2, regno2)

print("Hyundai car details:", Hyundai.getModel(), Hyundai.getRegno())
print("Maruthi car details:", Maruthi.getModel(), Maruthi.getRegno())
```

```python
# Python program to demonstrate method overriding 
# Defining parent class 
class Parent(): 
	# Constructor 
	def __init__(self): 
		self.myvalue1 = "Inside Parent"
		
	# Parent's show method 
	def show(self): 
		print(self.myvalue1) 

# Defining child class 
class Child(Parent): 
	def __init__(self): 
		self.value1 = "Inside Child"
		
	# Child's show method 
	
# Driver's code 
obj1 = Parent() 
obj2 = Child() 

obj1.show() 
obj2.show() 
```

```python
class Complex():
    def display(self):
        print(self.realPart,"+",self.imgPart,"i", sep="")
       
    def sum(self, c1, c2):
        self.realPart = c1.realPart + c2.realPart
        self.imgPart = c1.imgPart + c2.imgPart
    def initComplex(self):
        self.realPart = int(input("Real Part: "))
        self.imgPart = int(input("Imaginary Part: "))    
```

```python
c1 = Complex()
c2 = Complex()
c3 = Complex()
```

```python
print("complex number 1")
c1.initComplex()
c1.display()
print("complex number 2")
c2.initComplex()
c2.display()
c3.sum(c1,c2)
c3.display()
```

```python
# Python program to demonstrate overriding in multiple inheritance 
# Defining parent class 1 
class Parent1(): 
	# Parent's show method 
	def show(self): 
		print("Inside Parent1") 

# Defining Parent class 2 
class Parent2(): 
		
	# Parent's show method 
	def display(self): 
		print("Inside Parent2") 
		
      
# Defining child class 
class Child(Parent1, Parent2): 
	# Child's show method 
	def show(self): 
		print("Inside Child") 

        
# Driver's code 
obj = Child() 
obj.show() 
obj.display()
```

```python
# Python program to show that the variables with a value 
# assigned in class declaration, are class variables

# Class for Computer Science Student
class CSStudent:
	stream = 'cse'				 # Class Variable
	def __init__(self,name,roll):
		self.name = name		 # Instance Variable
		self.roll = roll		 # Instance Variable

        
# Objects of CSStudent class
a = CSStudent('Aryan', 1)
b = CSStudent('Jay', 2)
print(a.stream) # prints "cse"
print(b.stream) # prints "cse"
print(a.name) # prints "Aryan"
print(b.name) # prints "Jay"
print(a.roll) # prints "1"
print(b.roll) # prints "2"

print(CSStudent.stream) # prints "cse"

# To change the stream for all instances of the class we can change it 
# directly from the class
CSStudent.stream = 'Mechanical'

print(CSStudent.stream) 
```
